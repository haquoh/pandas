```python
# ━━━━ SQL 세팅 (코랩 첫 셀에 실행!) ━━━━

# 1. 구글 드라이브 연결
from google.colab import drive
drive.mount('/content/drive')

# 2. 필요한 도구 불러오기
import pandas as pd
import sqlite3              # SQLite — 파이썬에 기본 내장!

# 3. 데이터 경로
PATH = '/content/drive/MyDrive/Colab Notebooks/data/'

# 4. 데이터베이스 만들기 (파일 하나로!)
conn = sqlite3.connect('titanic.db')
# sqlite3.connect() = "titanic.db라는 데이터베이스를 만들거나 열어줘"
# conn = connection(연결)의 줄임말

# 5. 타이타닉 CSV를 데이터베이스 테이블로 넣기
df = pd.read_csv(PATH + 'titanic_train.csv')
df.to_sql('passengers', conn, index=False, if_exists='replace')
# 'passengers'라는 이름의 테이블로 저장
# if_exists='replace': 이미 있으면 덮어쓰기

print(f'✅ passengers 테이블 생성 완료! ({len(df)}행)')
```

```python
# SQL을 실행하고 결과를 예쁘게 보여주는 함수
def sql(query):
    """SQL 쿼리를 실행하고 결과를 DataFrame으로 반환"""
    result = pd.read_sql(query, conn)
    return result

# 사용법: sql("SQL 쿼리")
sql("SELECT * FROM passengers LIMIT 5")
```

```python
conn.execute("""
CREATE TABLE IF NOT EXISTS products (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    price INTEGER,
    category TEXT
)
""")
conn.commit()
print('✅ products 테이블 생성 완료')
```

```python
# 좀 더 복잡한 예시
conn.execute("""
CREATE TABLE IF NOT EXISTS employees (
    emp_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    department TEXT DEFAULT '미정',
    salary INTEGER,
    hire_date TEXT,
    is_active INTEGER DEFAULT 1
)
""")
conn.commit()
# DEFAULT '미정' = 부서를 안 넣으면 자동으로 '미정'
# DEFAULT 1 = 활성 여부를 안 넣으면 자동으로 1(활성)
```
